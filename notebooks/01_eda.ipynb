# 01 — Exploratory Data Analysis
Telco customer churn dataset.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('../ai/data/raw/telco_churn.csv')
df.head()
```

```python
df.info()
df.describe()
```

## Churn balance

```python
df['Churn'].value_counts(normalize=True).plot(kind='bar', title='Churn rate')
```

## Churn by contract type

```python
pd.crosstab(df['Contract'], df['Churn'], normalize='index').plot(kind='bar', stacked=True)
```
